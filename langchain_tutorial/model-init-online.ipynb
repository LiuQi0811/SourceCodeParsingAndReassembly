# 大数据模型
## 1. 调用具体模型厂商的API
### 1.1 调用 DeepSeek 官网的DeepSeek模型
调用大模型涉及到三个特别重要的参数：
base_url,model_name,api_key

将参数信息存储在.env配置文件中



```python
import os
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
from langchain_deepseek import ChatDeepSeek

# 1. 读取 .env 配置文件中的信息，相关的环境变量以 .env文件中的优先
load_dotenv(override=True)
DEEPSEEK_MODEL = os.getenv('DEEPSEEK_MODEL')
DEEPSEEK_API_KEY = os.getenv('DEEPSEEK_API_KEY')
DEEPSEEK_BASE_URL = os.getenv('DEEPSEEK_BASE_URL')
# 2. 模型的初始化
llm_deepseek = ChatDeepSeek(
                    model=DEEPSEEK_MODEL,
                     api_key=DEEPSEEK_API_KEY,
                     api_base=DEEPSEEK_BASE_URL
                     )
# 3.模型的调用
response= llm_deepseek.invoke('请用一句话介绍你自己')
print(response)
```

### 1.2 调用 AgnesAi 官网的模型
调用大模型涉及到三个特别重要的参数：
base_url,model_name,api_key
#### OpenAi 兼容所有的模型 （可以使用ChatOpenAI）
将参数信息存储在.env配置文件中

```python
import os
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI

# 1. 读取 .env 配置文件中的信息，相关的环境变量以 .env文件中的优先
load_dotenv(override=True)
AGNES_AI_API_KEY = os.getenv('AGNES_AI_API_KEY')
AGNES_AI_BASE_URL = os.getenv('AGNES_AI_BASE_URL')
AGNES_AI_MODEL = os.getenv('AGNES_AI_MODEL')
# 2. 模型的初始化
llm_openai = ChatOpenAI(
    model=AGNES_AI_MODEL,
    api_key=AGNES_AI_API_KEY,
    base_url=AGNES_AI_BASE_URL
)
# 3.模型的调用
response = llm_openai.invoke('请用一句话介绍你自己')
print(response)

```

# 2. 使用中转平台
## 2.1 使用 OpenRouter

```python
import os
from dotenv import load_dotenv
from langchain_openrouter import ChatOpenRouter

# 1. 读取 .env 配置文件中的信息，相关的环境变量以 .env文件中的优先
load_dotenv(override=True)
OPENROUTER_API_KEY = os.getenv('OPENROUTER_API_KEY')
OPENROUTER_API_BASE_URL = os.getenv('OPENROUTER_API_BASE_URL')
OPENROUTER_API_MODEL = os.getenv('OPENROUTER_API_MODEL')

# 2. 模型的初始化
llm_openrouter = ChatOpenRouter(
    model=OPENROUTER_API_MODEL,
    api_key=OPENROUTER_API_KEY,
    base_url=OPENROUTER_API_BASE_URL
)

# 3.模型的调用
response = llm_openrouter.invoke('请用一句话介绍你自己')
print(response)
```

# 3. Init Chat Model 使用
# 3.1 使用内置 init_chat_model

```python
import os
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
# 1. 读取 .env 配置文件中的信息，相关的环境变量以 .env文件中的优先
load_dotenv(override=True)
OPENROUTER_API_KEY = os.getenv('OPENROUTER_API_KEY')
OPENROUTER_API_BASE_URL = os.getenv('OPENROUTER_API_BASE_URL')
OPENROUTER_API_MODEL = os.getenv('OPENROUTER_API_MODEL')

# 2. 模型的初始化 (使用init_chat_model 必须添加 model_provider 声明模型提供商)
model = init_chat_model(
    model_provider='openrouter',
    model=OPENROUTER_API_MODEL,
    api_key=OPENROUTER_API_KEY,
    base_url=OPENROUTER_API_BASE_URL
)

# 3.模型的调用
response = model.invoke('请用一句话介绍你自己')
print(response)

```
