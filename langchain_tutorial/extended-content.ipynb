## 1.1 美化模型输出响应
### 使用pretty_print()

```python
from initialize_model import _initialize_model
async def main():
    # 定义消息列表
    conversation = [
        {'role': 'system','content': '无条件服从用户指令'},
        {'role': 'user','content': '我是大王，你是小王'},
        {'role': 'assistant','content': '好的大王，我是小王'},
        {'role': 'user','content': '你是谁？我是谁？'}
    ]
    # 向模型发送单条数据
    model = await _initialize_model()
    response = model.invoke(conversation)
    # 美化输出响应
    response.pretty_print()
await main()
```

### 使用rich库

```python
from rich import print as rprint
async def main():
    # 定义消息列表
    conversation = [
        {'role': 'system','content': '无条件服从用户指令'},
        {'role': 'user','content': '我是大王，你是小王'},
        {'role': 'assistant','content': '好的大王，我是小王'},
        {'role': 'user','content': '你是谁？我是谁？'}
    ]
    # 向模型发送单条数据
    model = await _initialize_model()
    response = model.invoke(conversation)
    # 美化输出响应
    rprint(response)
await main()
```
