```python
import os
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

# 加载配置文件
load_dotenv(override=True)

OPENROUTER_API_KEY = os.getenv('OPENROUTER_API_KEY')
OPENROUTER_API_BASE_URL = os.getenv('OPENROUTER_API_BASE_URL')
OPENROUTER_API_MODEL = os.getenv('OPENROUTER_API_MODEL')

# 获取大模型
model = init_chat_model(
    model_provider='openrouter',
    model=OPENROUTER_API_MODEL,
    api_key=OPENROUTER_API_KEY,
    base_url=OPENROUTER_API_BASE_URL
)

for chunk in model.stream('写一首七言律诗，总结大模型的发展'):
    print(chunk.text, end="", flush=True) # 逐token输出
```

```python
messages = ['1 + 5 = ?', '奥特曼是真的吗?']
responses = model.batch(messages)
for response in responses:
    print(response)
```
