# 模型调用的 invoke使用
## 1. invoke的传参
### 1.1 文本输入

```python
import os
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

# 加载配置文件
load_dotenv(override=True)

OPENROUTER_API_KEY = os.getenv('OPENROUTER_API_KEY')
OPENROUTER_API_BASE_URL = os.getenv('OPENROUTER_API_BASE_URL')
OPENROUTER_API_MODEL = os.getenv('OPENROUTER_API_MODEL')

# 2. 获取大模型 (使用init_chat_model 必须添加 model_provider 声明模型提供商)
model = init_chat_model(
    model_provider='openrouter',
    model=OPENROUTER_API_MODEL,
    api_key=OPENROUTER_API_KEY,
    base_url=OPENROUTER_API_BASE_URL
)

# 3.模型的调用
response = model.invoke('翻译如下的内容： 你好 我的好朋友')
print(response.content)
```

### 1.2 字典列表

```python

messages = [
    {'role': 'system','content':'你是一个专业的数学老师'},
    {'role': 'user', 'content': '帮我解释一下什么是斐波那契数列'}
]
response = model.invoke(messages)
print(response.content)
```

```python
messages = [
    {'role': 'system','content':'你是一个专业的数学老师'},
    {'role': 'user', 'content': '1 + 2 = ?'},
    {'role': 'assistant','content': '3'},
    {'role': 'user', 'content':'我刚才问了什么问题'}
]
response = model.invoke(messages)
print(response.content)
```

```python
messages = [
    {'role': 'system','content':'你是一个非常友好的AI助手'},
    {'role': 'user', 'content': '你好，我叫云悠悠'}
]
# 第一个对话
response = model.invoke(messages)
print(f'AI的第一次回复：{response.content}')
messages_ = [
    {'role': 'user', 'content': '我叫什么名字?'}
]
# 第二次对话
response_ = model.invoke(messages_)
print(f'AI的第二次回复：{response_.content}')
```

```python
conversation = [
    {'role': 'system','content':'你是一个非常友好的AI助手'},
    {'role': 'user', 'content': '你好，我叫云悠悠'}
]
# 第一个对话
response = model.invoke(conversation)
print(f'AI的第一次回复：{response.content}')

# 添加记忆
conversation.append({
    'role': 'assistant',
    'content': response.content
})
conversation.append({
    'role': 'user',
    'content': '我叫什么名字?'
})
# 第二次对话
response_ = model.invoke(conversation)
print(f'AI的第二次回复：{response_.content}')
```

### 1.3 消息对象列表


```python
from langchain_core.messages import SystemMessage,HumanMessage
messages = [
    SystemMessage(content='你是一个专业的UI设计师'),
    HumanMessage(content='帮我分析一下苹果官网设计'),
]
response = model.invoke(messages)
print(response.content)
```

```python
from langchain_core.messages import SystemMessage,HumanMessage,AIMessage
messages = [
    SystemMessage(content='你是一个专业的UI设计师'),
    HumanMessage(content='苹果官网设计简单吗?'),
]
# 第一次对话
response = model.invoke(messages)
print(f'AI 第一次回复：{response.content}')

messages = [
    AIMessage(content=response.content),
    HumanMessage(content='我刚才问了什么问题?')
]
# 第二次对话
response_ = model.invoke(messages)
print(f'AI 第二次回复：{response_.content}')
```

```python
response = model.invoke('请解释 1024')
# 1. 获取回复内容
print('AI 回复: ',response.content)

# 2. 获取响应元数据
metadata = response.response_metadata
print(f'使用的模型: {metadata['model_name']}')
print(f"结束原因: {metadata['finish_reason']}")
print(f"模型提供商：{metadata['model_provider']}\n")

# 3. 获取 Token 使用情况
usage = metadata.get('token_usage', {})
print(f"输入 tokens: {usage.get('prompt_tokens')}")
print(f"输出 tokens: {usage.get('completion_tokens')}")
print(f"总计 tokens: {usage.get('total_tokens')}")

# 4. 获取消息 ID
print(f"消息 ID: {response.id}")
```
