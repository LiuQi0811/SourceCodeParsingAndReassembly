# 模型初始化参数的使用情况
## 1. temperature的使用


```python
import os
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
# 1. 读取 .env 配置文件中的信息，相关的环境变量以 .env文件中的优先
load_dotenv(override=True)
AGNES_AI_API_KEY = os.getenv('AGNES_AI_API_KEY')
AGNES_AI_BASE_URL = os.getenv('AGNES_AI_BASE_URL')
AGNES_AI_MODEL = os.getenv('AGNES_AI_MODEL')

# 2. 模型的初始化 (使用init_chat_model 必须添加 model_provider 声明模型提供商)
model = init_chat_model(
    model_provider='openai',
    model=AGNES_AI_MODEL,
    temperature=0,
    api_key=AGNES_AI_API_KEY,
    base_url=AGNES_AI_BASE_URL
)

# 3.模型的调用
response = model.invoke('张胜男, 男 , 25岁，帮我从上文提取数据返回json 格式')
print(response.content)
```

## 2.max_tokens的使用

```python
import os
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
# 1. 读取 .env 配置文件中的信息，相关的环境变量以 .env文件中的优先
load_dotenv(override=True)
AGNES_AI_API_KEY = os.getenv('AGNES_AI_API_KEY')
AGNES_AI_BASE_URL = os.getenv('AGNES_AI_BASE_URL')
AGNES_AI_MODEL = os.getenv('AGNES_AI_MODEL')

# 2. 模型的初始化 (使用init_chat_model 必须添加 model_provider 声明模型提供商)
model = init_chat_model(
    model_provider='openai',
    model=AGNES_AI_MODEL,
    api_key=AGNES_AI_API_KEY,
    base_url=AGNES_AI_BASE_URL,
    max_tokens=10
)

# 3.模型的调用
response = model.invoke('介绍一下你自己')
print(response)
```
